```python
##Q7.5
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside function:")
    print("val1 =", val1)
    print("val2 =", val2)

x = 6
y = 3

swap_message(x, y)

print("Outside function:")
print("x =", x)
print("y =", y)
```

```python
##Q8.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(val)
```

val is a parameter (local variable) of the function my_abs().
It exists only inside the function.
print(val) is written outside the function, where val does not exist.
Therefore, Python cannot find val and raises a NameError.

```python
##Q8.2
counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
##Q8.4
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
