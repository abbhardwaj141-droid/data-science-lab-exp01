```python
##Q35
np.random.seed(42)

marks = np.random.randint(20, 101, (12, 4))

print("Student marks:\n", marks)

averages = marks.mean(axis=1)
print("Student averages:", averages)

print("Test means:", marks.mean(axis=0))
print("Test standard deviations:", marks.std(axis=0))

passed = np.all(marks >= 40, axis=1)
print("Students passing all tests:", np.sum(passed))

print("Students with average >= 75:", np.where(averages >= 75)[0] + 1)

print("Test with greatest variation:", np.argmax(marks.std(axis=0)) + 1)
```

```python
##Q36
a = np.arange(21, 46).reshape(5, 5)
print("Original matrix:\n", a)
print("Original shape:", a.shape)

b = a[1:4, 1:4]
print("Central matrix:\n", b)
print("Central shape:", b.shape)

c = b.T
print("Transpose:\n", c)
print("Transpose shape:", c.shape)

d = np.dot(b, c)
print("Matrix product:\n", d)
print("Product shape:", d.shape)

row_sums = d.sum(axis=1)
print("Row sums:", row_sums)

maximum = row_sums.max()
print("Maximum row sum:", maximum)
print("Positions with maximum:", np.where(row_sums == maximum)[0])
```

```python
##Q37
quantities = np.array([
    [10, 12, 15, 8],
    [14, 10, 18, 12],
    [20, 15, 10, 9],
    [12, 18, 14, 16],
    [16, 20, 12, 10],
    [18, 14, 16, 15]
])

prices = np.array([150, 240, 90, 320])

daily_revenue = quantities @ prices
product_quantity = quantities.sum(axis=0)
product_revenue = quantities.sum(axis=0) * prices

print("Daily revenue:", daily_revenue)
print("Total revenue per product:", product_revenue)
print("Total revenue:", daily_revenue.sum())
print("Highest revenue day:", np.argmax(daily_revenue) + 1)
print("Best-selling product:", np.argmax(product_quantity) + 1)

new_prices = np.where(product_quantity > 100, prices * 0.90, prices)

print("Original prices:", prices)
print("Discounted prices:", new_prices)
print("Revenue after discount:", quantities @ new_prices)
print("Total revenue after discount:", (quantities @ new_prices).sum())
```

```python
##Q38
np.random.seed(42)

data = np.random.normal(50, 12, 100)

cleaned = np.where(data < 30, 30, np.where(data > 70, 70, data))

print("Values outside 30-70:", np.sum((data < 30) | (data > 70)))

print("\nOriginal statistics:")
print("Mean:", np.mean(data))
print("Variance:", np.var(data))
print("Standard deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))

print("\nCleaned statistics:")
print("Mean:", np.mean(cleaned))
print("Variance:", np.var(cleaned))
print("Standard deviation:", np.std(cleaned))
print("Minimum:", np.min(cleaned))
print("Maximum:", np.max(cleaned))
```

```python
##Q39
branch_a = pd.Series({"Q1": 120, "Q2": 145, "Q4": 180})
branch_b = pd.Series({"Q2": 110, "Q3": 135, "Q4": 155})
branch_c = pd.Series({"Q1": 95, "Q3": 125, "Q4": 140})

result = branch_a + branch_b + branch_c

print("Ordinary addition:\n", result)
print("Missing values:\n", result[result.isnull()])

common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
print("Common quarters:", common)

print("Combined earnings for common quarters:")
print(branch_a[common] + branch_b[common] + branch_c[common])

zero_result = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print("Missing records treated as zero:\n", zero_result)
```

```python
##Q40
scores = {
    "R101": 78,
    "R102": 65,
    "R104": 92,
    "R105": 48
}

required = ["R105", "R103", "R101", "R106", "R104", "R102"]

s = pd.Series(scores, index=required)

print("Original Series:\n", s)
print("Missing student records:\n", s[s.isnull()])

filtered = s[s.notnull() & (s >= 70)]
print("Valid scores >= 70:\n", filtered)

d = filtered.to_dict()
print("Dictionary:", d)

new_series = pd.Series(d)
print("Reconstructed Series:\n", new_series)

print("Values match:", filtered.equals(new_series))
```
