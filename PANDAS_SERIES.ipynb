```python
import pandas as pd

sales = pd.Series(
    [600, 450, 800, 700],
    index=['Pen', 'Book', 'Bag', 'Shoes']
)

# Display values
print(sales.values)

# Display index
print(sales.index)

# Select one label
print(sales['Bag'])

# Filter sales above 500
print(sales[sales > 500])

# Check whether label exists
print('Pen' in sales.index)

# Convert Series to dictionary
sales_dict = sales.to_dict()

print(sales_dict)
```
